```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
# Student marks dataset-EXAMPLE 2
data = {
    'Maths': [87, 75, 91, 45, 50, 35, 78, 82, 39, 46],
    'Science': [82, 73, 91, 40, 48, 40, 92, 80, 35, 44]
}

df = pd.DataFrame(data)

# Apply K-Means
kmeans = KMeans(n_clusters=2, random_state=42)

df['Cluster'] = kmeans.fit_predict(df)

# Display dataset with clusters
print(df)

# Plot the clusters
plt.scatter(df['Maths'], df['Science'],
            c=df['Cluster'],
            cmap='viridis',
            s=100)

# Plot centroids
plt.scatter(kmeans.cluster_centers_[:,0],
            kmeans.cluster_centers_[:,1],
            color='red',
            marker='X',
            s=250,
            label='Centroids')

plt.xlabel("Maths Marks")
plt.ylabel("Science Marks")
plt.title("K-Means Clustering of Student Marks")
plt.legend()
plt.grid(True)
plt.show()
```

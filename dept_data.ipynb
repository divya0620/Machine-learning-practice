```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame({
    'Roll_No': [201, 202, 203, 204, 205, 206, 207, 208, 209, 210, 211],
    'Name': ['Liam', 'Olivia', 'Noah', 'Emma', 'Ava',
             'William', 'Sophia', 'James', 'Isabella', 'Mason', 'Mia'],

    'Department': ['CSE', 'CSE', 'HR', 'HR', 'CSE',
                   'Finance', 'Finance', 'HR', 'CSE', 'Finance', 'HR'],

    'Age': [19, 20, 21, 22, 20, 23, 21, 22, 19, 24, 20],
    'Math': [78, 85, 88, 90, 76, 84, 79, 92, 87, 81, 75],
    'Science': [82, 89, 91, 86, 80, 88, 83, 94, 90, 85, 77]
})

print(df)
```

```python
df['Total'] = df['Math'] + df['Science']
print(df)
```

```python
df.drop('Science', axis=1, inplace=True)

print(df)
```

```python
pivot_dept = df.pivot_table(
    index='Department',
    values='Total',
    aggfunc='mean'
)

print(pivot_dept)
```

```python
df.groupby('Department')['Total'].mean()
```

```python
df_missing = df.copy()

df_missing.loc[2, 'Math'] = np.nan
df_missing.loc[7, 'Total'] = np.nan

print(df_missing)

df_missing.isnull()
```

```python
df.loc[0:4, ['Name', 'Department', 'Total']]

df.loc[df['Department'] == 'CSE']
```

```python
grouped = df.groupby('Department')

grouped['Total'].mean()
grouped['Math'].sum()
```

```python
multi_index_df = df.groupby(['Department', 'Name'])['Total'].mean()
print(multi_index_df)
```
